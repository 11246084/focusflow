```python
import os

input_dir = "images/" # 定義輸入圖像文件夾路徑
target_dir = "annotations/trimaps/" # 定義目標標註文件夾路徑

# 獲取所有 .jpg 圖像文件的路徑，並按名稱排序
input_img_paths = sorted(
    [os.path.join(input_dir, fname)
     for fname in os.listdir(input_dir)
     if fname.endswith(".jpg")])
# 獲取所有 .png 標註文件的路徑，排除以 '.' 開頭的文件，並按名稱排序
target_paths = sorted(
    [os.path.join(target_dir, fname)
     for fname in os.listdir(target_dir)
     if fname.endswith(".png") and not fname.startswith(".")])
```

```python
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array

def display_target(target_array):
    # 將目標數組歸一化到0-255範圍，便於顯示。原始標註值為1,2,3，轉換後為0,127,254
    normalized_array = (target_array.astype("uint8") - 1) * 127
    plt.axis("off") # 關閉座標軸顯示
    plt.imshow(normalized_array[:, :, 0]) # 顯示處理後的標註圖像 (灰度圖只需一個通道)

# 載入第10張標註圖像 (索引為9)，並轉換為灰度圖數組
img = img_to_array(load_img(target_paths[9], color_mode="grayscale"))
display_target(img) # 顯示標註圖像
```

```python
import numpy as np
import random
from tensorflow.keras.utils import load_img, img_to_array # 確保所有必要庫都已導入

img_size = (200, 200) # 定義圖像尺寸 (寬, 高)
num_imgs = len(input_img_paths) # 獲取圖像總數

random.Random(1337).shuffle(input_img_paths) # 使用固定種子打亂輸入圖像路徑，確保每次運行結果一致
random.Random(1337).shuffle(target_paths) # 使用固定種子打亂目標標註路徑，與輸入圖像保持同步

def path_to_input_image(path):
    # 載入輸入圖像並調整大小，返回圖像數組
    return img_to_array(load_img(path, target_size=img_size))

def path_to_target(path):
    # 載入目標標註圖像，調整大小，轉為灰度圖數組
    img = img_to_array(
        load_img(path, target_size=img_size, color_mode="grayscale"))
    # 將標註值從1,2,3轉換為0,1,2，作為類別標籤 (方便進行稀疏類別交叉熵計算)
    img = img.astype("uint8") - 1
    return img

# 初始化用於存儲輸入圖像和目標標註的NumPy數組
# input_imgs: (樣本數, 高, 寬, 顏色通道數)
input_imgs = np.zeros((num_imgs,) + img_size + (3,), dtype="float32")
# targets: (樣本數, 高, 寬, 1) 因為是單通道的類別標籤
targets = np.zeros((num_imgs,) + img_size + (1,), dtype="uint8")

# 遍歷所有圖像路徑，載入並預處理圖像和標註
for i in range(num_imgs):
    input_imgs[i] = path_to_input_image(input_img_paths[i])
    targets[i] = path_to_target(target_paths[i])

num_val_samples = 1000 # 定義驗證集的樣本數量
# 劃分訓練集和驗證集，最後 num_val_samples 個樣本用於驗證
train_input_imgs = input_imgs[:-num_val_samples]
train_targets = targets[:-num_val_samples]
val_input_imgs = input_imgs[-num_val_samples:]
val_targets = targets[-num_val_samples:]
```

```python
from tensorflow import keras
from tensorflow.keras import layers

def get_model(img_size, num_classes):
    # 定義模型輸入層，形狀為圖像尺寸加上3個顏色通道，輸入的shape為(200,200,3)
    inputs = keras.Input(shape=img_size + (3,))
    # 將輸入圖像像素值縮放到0-1範圍 (標準化)
    x = layers.Rescaling(1./255)(inputs)

    # 編碼器部分 (降採樣路徑)，我們在每個Conv2D層都用了[padding="same"]，藉此避免邊界效應對特徵圖的尺寸造成影響
    x = layers.Conv2D(64, 3, strides=2, activation="relu", padding="same")(x) # 卷積層，步長為2來縮小採樣數
    x = layers.Conv2D(64, 3, activation="relu", padding="same")(x) # 普通卷積層
    x = layers.Conv2D(128, 3, strides=2, activation="relu", padding="same")(x)
    x = layers.Conv2D(128, 3, activation="relu", padding="same")(x)
    x = layers.Conv2D(256, 3, strides=2, padding="same", activation="relu")(x)
    x = layers.Conv2D(256, 3, activation="relu", padding="same")(x)

    # 解碼器部分 (升採樣路徑)
    x = layers.Conv2DTranspose(256, 3, activation="relu", padding="same")(x) # 反卷積層，用於上採樣
    x = layers.Conv2DTranspose(256, 3, activation="relu", padding="same", strides=2)(x) # 步長為2來縮小採樣數
    x = layers.Conv2DTranspose(128, 3, activation="relu", padding="same")(x)
    x = layers.Conv2DTranspose(128, 3, activation="relu", padding="same", strides=2)(x)
    x = layers.Conv2DTranspose(64, 3, activation="relu", padding="same")(x)
    x = layers.Conv2DTranspose(64, 3, activation="relu", padding="same", strides=2)(x)

    # 輸出層，使用 softmax 激活函數進行多類別像素分類
    # num_classes 是分割的類別數量 (例如，背景、前景、邊緣)
    outputs = layers.Conv2D(num_classes, 3, activation="softmax", padding="same")(x)

    # 創建 Keras 模型，指定輸入和輸出
    model = keras.Model(inputs, outputs)
    return model

# 實例化模型，傳入圖像尺寸和類別數量 (數據集中有3個類別：背景、動物、邊緣)
model = get_model(img_size=img_size, num_classes=3)
model.summary() # 打印模型結構摘要，查看各層輸出形狀和參數數量
```

```python
model.compile(optimizer="rmsprop",loss="sparse_categorical_crossentropy") # 損失函數為稀疏類別交叉熵 (適用於整數標籤的分類問題)

callbacks = [
    keras.callbacks.ModelCheckpoint("oxford_segmentation.keras", # 保存模型的路徑和文件名
                    save_best_only=True) # 只保存驗證集上性能最好的模型權重
]

# 訓練模型
history = model.fit(train_input_imgs, train_targets,
           epochs=50, # 設定訓練的周期數
           callbacks=callbacks, # 使用定義的回調函數列表
           batch_size=64, # 設定每次梯度更新的樣本數
           validation_data=(val_input_imgs, val_targets)) # 指定用於驗證的數據集
```

```python
# 獲取訓練的 epoch 範圍
epochs = range(1, len(history.history["loss"]) + 1)
# 獲取訓練損失
loss = history.history["loss"]
# 獲取驗證損失
val_loss = history.history["val_loss"]
plt.figure() # 創建新的圖形
plt.plot(epochs, loss, "bo", label="Training loss") # 繪製訓練損失 (藍色圓點)
plt.plot(epochs, val_loss, "b", label="Validation loss") # 繪製驗證損失 (藍色實線)
plt.title("Training and validation loss") # 圖形標題
plt.legend() # 顯示圖例
plt.show() # 顯示圖形
```

可以看到在訓練中期(25個週期附近)就開始出現過度配適。

```python
from tensorflow.keras.utils import array_to_img
import numpy as np # 確保所有必要庫都已導入
import matplotlib.pyplot as plt # 確保所有必要庫都已導入
from tensorflow import keras # 確保 keras 已導入以使用 models.load_model

model = keras.models.load_model("oxford_segmentation.keras") # 載入保存的最佳模型 (基於驗證損失)

i = 4 # 選擇驗證集中的第5個圖像 (索引為4) 作為測試樣本
test_image = val_input_imgs[i] # 獲取測試輸入圖像
plt.axis("off") # 關閉座標軸顯示
plt.imshow(array_to_img(test_image)) # 顯示原始測試圖像
plt.title("Original Test Image") # 添加標題
plt.show() # 顯示圖像

# 使用模型對測試圖像進行預測，得到分割掩碼的概率分佈
# np.expand_dims 增加一個批次維度，因為模型期望批次輸入 (例如，(1, 200, 200, 3))
mask = model.predict(np.expand_dims(test_image, 0))[0]

def display_mask(pred):
    # 從模型的預測輸出中獲取每個像素的最高概率類別，作為分割掩碼
    # np.argmax 沿最後一個軸 (類別軸) 找到最大值的索引
    mask = np.argmax(pred, axis=-1)
    # 將類別值轉換為可視化的灰度值 (例如，0, 127, 254)，以便在灰度圖中區分不同類別
    mask *= 127
    plt.axis("off") # 關閉座標軸顯示
    plt.imshow(mask) # 顯示預測的分割掩碼
    plt.title("Predicted Segmentation Mask") # 添加標題
    plt.show() # 顯示圖像

display_mask(mask) # 顯示預測的分割掩碼
```
