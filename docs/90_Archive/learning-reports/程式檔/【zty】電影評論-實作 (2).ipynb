## **【程式8.1】**

初始化一個小型卷積神經網路模型

```python
# 導入必要的函式庫
from tensorflow import keras
from tensorflow.keras import layers

# 定義模型的輸入層，輸入圖像的尺寸為28(圖片高度)x28(圖片寬度)像素，1(channel)代表灰度圖像（單通道）
inputs = keras.Input(shape=(28,28,1))

# 第一個卷積層：使用32個3x3大小的濾波器，啟動函數為ReLU
x = layers.Conv2D(filters=32 , kernel_size=3, activation="relu")(inputs)

# 第一個最大池化層：將特徵圖的尺寸減半，池化窗口大小為2x2
x = layers.MaxPooling2D(pool_size = 2)(x)

# 第二個卷積層：使用64個3x3大小的濾波器，啟動函數為ReLU
x = layers.Conv2D(filters=64 , kernel_size=3, activation="relu")(x)

# 第二個最大池化層：將特徵圖的尺寸減半，池化窗口大小為2x2
x = layers.MaxPooling2D(pool_size = 2)(x)

# 第三個卷積層：使用128個3x3大小的濾波器，啟動函數為ReLU
x = layers.Conv2D(filters=128 , kernel_size=3, activation="relu")(x)

# 加入扁平層
# 將三維的特徵圖展平為一維向量，以便輸入到全連接層
x = layers.Flatten()(x)

# 輸出層：一個全連接層，有10個神經元（對應10個類別），啟動函數為Softmax用於多分類
outputs = layers.Dense(10, activation="softmax")(x)

# 建立Keras模型，指定輸入和輸出
model = keras.Model(inputs=inputs, outputs=outputs)
```

## **【程式8.3】**

使用MNIST數字來訓練卷積神經網路

```python
from tensorflow.keras.datasets import mnist

# 載入MNIST數據集，分為訓練圖像、訓練標籤、測試圖像、測試標籤
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 將訓練圖像的形狀從 (60000, 28, 28) 轉換為 (60000, 28, 28, 1)，增加一個通道維度（灰度圖為1）
train_images = train_images.reshape((60000, 28, 28, 1))
# 將訓練圖像的像素值從 0-255 範圍正規化到 0-1 範圍，便於模型訓練
train_images = train_images.astype('float32') / 255

# 將測試圖像的形狀從 (10000, 28, 28) 轉換為 (10000, 28, 28, 1)
test_images = test_images.reshape((10000, 28, 28, 1))
# 將測試圖像的像素值從 0-255 範圍正規化到 0-1 範圍
test_images = test_images.astype('float32') / 255

# 配置模型的編譯器：
# optimizer='rmsprop'：使用 RMSprop 優化器來調整模型權重。
# loss='sparse_categorical_crossentropy'：由於標籤是整數形式（非獨熱編碼），使用此損失函數進行多分類。
# metrics=['accuracy']：在訓練和評估期間，模型將報告準確度（accuracy）。
model.compile(optimizer='rmsprop',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 訓練模型：
# train_images：訓練用的圖像數據。
# train_labels：訓練圖像對應的真實標籤。
# epochs=5：模型將對整個訓練數據集進行 5 次迭代。
# batch_size=64：每次梯度更新使用的樣本數量為 64。
model.fit(train_images, train_labels, epochs=5, batch_size=64)
```

## **【程式8.4】**

用測試資料來評估模型，模型在測試數據上的表現不錯，最終的測試準確度為:99.24%。這表示模型在未曾見過的測試圖像上，有大約99.24%的機會能正確識別數字。

```python
# 使用測試數據評估模型，返回測試損失 (test_loss) 和測試準確度 (test_acc)
test_loss, test_acc = model.evaluate(test_images, test_labels)

# 印出測試準確度
print(f"Test accuracy: {test_acc:.3f}")
```
